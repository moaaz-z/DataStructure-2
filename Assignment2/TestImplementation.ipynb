```python
import random
import time
import matplotlib.pyplot as plt

sizes = [100, 500, 1000, 2000, 5000]

bubble_times = []
insertion_times = []
selection_times = []
merge_times = []
heap_times = []
quick_times = []
hybrid_times = []

for n in sizes:
    arr = [random.randint(1, 10000) for i in range(n)]

    # Create all copies before measuring
    bubble_arr = arr.copy()
    insertion_arr = arr.copy()
    selection_arr = arr.copy()
    merge_arr = arr.copy()
    heap_arr = arr.copy()
    quick_arr = arr.copy()
    hybrid_arr = arr.copy()

    # Bubble Sort
    start = time.perf_counter()
    s.bubble_sort(bubble_arr)
    bubble_times.append(time.perf_counter() - start)

    # Insertion Sort
    start = time.perf_counter()
    s.insertion_sort(insertion_arr)
    insertion_times.append(time.perf_counter() - start)

    # Selection Sort
    start = time.perf_counter()
    s.selection_sort(selection_arr)
    selection_times.append(time.perf_counter() - start)

    # Merge Sort
    start = time.perf_counter()
    s.merge_sort(merge_arr)
    merge_times.append(time.perf_counter() - start)

    # Heap Sort
    start = time.perf_counter()
    s.heap_sort(heap_arr)
    heap_times.append(time.perf_counter() - start)

    # Quick Sort
    start = time.perf_counter()
    s.quick_sort(quick_arr)
    quick_times.append(time.perf_counter() - start)

    # Hybrid Merge Sort
    start = time.perf_counter()
    s.hybrid_merge_sort(hybrid_arr, 32)
    hybrid_times.append(time.perf_counter() - start)


# Print exact measurements
print("Execution times in seconds:\n")

for i, n in enumerate(sizes):
    print(f"Array size: {n}")
    print(f"Bubble Sort:        {bubble_times[i]:.8f} s")
    print(f"Insertion Sort:     {insertion_times[i]:.8f} s")
    print(f"Selection Sort:     {selection_times[i]:.8f} s")
    print(f"Merge Sort:         {merge_times[i]:.8f} s")
    print(f"Heap Sort:          {heap_times[i]:.8f} s")
    print(f"Quick Sort:         {quick_times[i]:.8f} s")
    print(f"Hybrid Merge Sort:  {hybrid_times[i]:.8f} s")
    print()


# Plot the results
plt.figure(figsize=(12, 7))

plt.plot(sizes, bubble_times, marker='o', label='Bubble Sort')
plt.plot(sizes, insertion_times, marker='o', label='Insertion Sort')
plt.plot(sizes, selection_times, marker='o', label='Selection Sort')
plt.plot(sizes, merge_times, marker='o', label='Merge Sort')
plt.plot(sizes, heap_times, marker='o', label='Heap Sort')
plt.plot(sizes, quick_times, marker='o', label='Quick Sort')
plt.plot(sizes, hybrid_times, marker='o', label='Hybrid Merge Sort')

plt.xlabel('Array Size (n)')
plt.ylabel('Execution Time (seconds)')
plt.title('Sorting Algorithms Comparison')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
