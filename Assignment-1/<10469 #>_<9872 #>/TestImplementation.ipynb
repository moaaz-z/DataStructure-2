## Generate Array

```python
import random
import SortingAlgorithms as s
import time
def generate_random_array(size, lower_bound, upper_bound):
    return [random.randint(lower_bound, upper_bound) for _ in range(size)]



```

Bubble Sort

```python
arr1 = generate_random_array(10000, 1, 10000)
start=time.process_time()
sorted_array = s.bubble_sort(arr1)
end=time.process_time()
print("Bubble Sort Time is ", end-start,"ms")
```

Selection Sort

```python
arr2 = generate_random_array(10000, 1, 10000)
start=time.process_time()
sorted_array = s.selection_sort(arr2)
end=time.process_time()
print("Selection Sort Time is ", end-start,"ms")
```

Insertion Sort

```python
arr3 = generate_random_array(10000, 1, 10000)
start=time.process_time()
sorted_array = s.insertion_sort(arr3)
end=time.process_time()
print("Insertion Sort Time is ", end-start,"ms")
```

```python

import matplotlib.pyplot as plt


sizes = [100, 500, 1000, 2000, 5000]

bubble_times = []
insertion_times = []
selection_times = []

for n in sizes:
    arr = [random.randint(1, 10000) for i in range(n)]

    start = time.perf_counter()
    s.bubble_sort(arr.copy())
    bubble_times.append(time.perf_counter() - start)

    start = time.perf_counter()
    s.insertion_sort(arr.copy())
    insertion_times.append(time.perf_counter() - start)

    start = time.perf_counter()
    s.selection_sort(arr.copy())
    selection_times.append(time.perf_counter() - start)


plt.plot(sizes, bubble_times, marker='o', label='Bubble Sort')
plt.plot(sizes, insertion_times, marker='o', label='Insertion Sort')
plt.plot(sizes, selection_times, marker='o', label='Selection Sort')

plt.xlabel('Array Size (n)')
plt.ylabel('Execution Time (seconds)')
plt.title('Sorting Algorithms Comparison')

plt.legend()
plt.grid(True)
plt.show()
```
